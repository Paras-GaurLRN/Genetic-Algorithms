```python
import numpy as np
import pandas as pd
import pygad as pgd
```

```python
def fitness_func(ga_instance,solution,solution_idx):
    return min(-(np.sum(solution*np.asarray([2,3,-0.5,-1,0.3],dtype=np.float64))-32),(np.sum(solution*np.asarray([2,3,-0.5,-1,0.3],dtype=np.float64))-32))
```

```python
ga = pgd.GA(
    num_generations=10000,
    num_parents_mating=200,
    sol_per_pop=1000,
    num_genes=5,
    fitness_func=fitness_func,
    mutation_type=None,
    random_seed=34
)
```

```python
ga.run()
```

```python
sol, sol_fitness, sol_idx = ga.best_solution()
```

```python
print("Weights : ",sol)
print("Fitness : ",sol_fitness)
print("Value : ",np.sum(sol*np.asarray([2,3,-0.5,-1,0.3],dtype=np.float64)))
```
